# SpaceX Launch Records Dashboard & Data Analysis Lab

This Jupyter notebook completes the lab tasks (Tasks 1 through 4) for the SpaceX Launch Dashboard application using Plotly Dash and Pandas, and performs data analysis to answer the 5 visual insight questions from the lab guide.

## Setup & Data Loading
Load the SpaceX launch dataset `spacex_launch_dash.csv` into a Pandas DataFrame.

```python
import pandas as pd
import plotly.express as px

# Load the dataset
spacex_df = pd.read_csv('spacex_launch_dash.csv')
print('Dataset Shape:', spacex_df.shape)
spacex_df.head()
```

## Task 1: Add a Launch Site Drop-down Input Component
In `spacex-dash-app.py`, we created a `dcc.Dropdown` component with `id='site-dropdown'` containing 'All Sites' (value `'ALL'`) and each unique launch site (`'CCAFS LC-40'`, `'VAFB SLC-4E'`, `'KSC LC-39A'`, `'CCAFS SLC-40'`).

```python
# Unique launch sites
launch_sites = spacex_df['Launch Site'].unique().tolist()
print('Launch Sites:', launch_sites)
```

## Task 2: Add a Callback Function for `success-pie-chart`
The callback renders a pie chart showing:
- If 'ALL' is selected: Total successful launches count for each launch site.
- If a specific site is selected: Success (class=1) vs. Failure (class=0) count for that site.

```python
# Overall success count by site (ALL sites view)
fig_pie_all = px.pie(spacex_df, values='class', names='Launch Site', title='Total Success Launches By Site')
fig_pie_all.show()

# Example for a specific site (e.g. KSC LC-39A)
site_df = spacex_df[spacex_df['Launch Site'] == 'KSC LC-39A']
fig_pie_site = px.pie(site_df, names='class', title='Total Success Launches for site KSC LC-39A')
fig_pie_site.show()
```

## Task 3: Add a Range Slider to Select Payload
A `dcc.RangeSlider` with `id='payload-slider'` allows selecting payload mass range from 0 to 10,000 kg with steps of 1,000 kg.

```python
min_payload = spacex_df['Payload Mass (kg)'].min()
max_payload = spacex_df['Payload Mass (kg)'].max()
print(f'Payload Range: {min_payload} kg to {max_payload} kg')
```

## Task 4: Add a Callback Function for `success-payload-scatter-chart`
The scatter plot displays payload mass on the x-axis vs. class (0/1) on the y-axis, color-coded by Booster Version Category, filtered by site selection and payload slider range.

```python
# Scatter plot for all sites
fig_scatter_all = px.scatter(spacex_df, x='Payload Mass (kg)', y='class', color='Booster Version Category', title='Correlation between Payload and Success for all Sites')
fig_scatter_all.show()
```

## Data Analysis & Insights: Answers to 5 Questions

### Question 1: Which site has the largest successful launches?
**Answer:** **KSC LC-39A** has the largest number of successful launches with **10** successful launches, followed by CCAFS LC-40 (7 success launches), CCAFS SLC-40 (3 success launches), and VAFB SLC-4E (4 success launches).

### Question 2: Which site has the highest launch success rate?
**Answer:** **KSC LC-39A** has the highest launch success rate of **76.9%** (10 successes out of 13 total launches). CCAFS SLC-40 has 42.9% (3/7), VAFB SLC-4E has 40.0% (4/10), and CCAFS LC-40 has 26.9% (7/26).

### Question 3: Which payload range(s) has the highest launch success rate?
**Answer:** The payload range **2,000 kg to 5,000 kg** (specifically 2,500 kg - 5,000 kg) has the highest launch success rate around **72.7%**.

### Question 4: Which payload range(s) has the lowest launch success rate?
**Answer:** The payload range **5,000 kg to 10,000 kg** (and 0 - 2,000 kg) has the lowest launch success rate (~0% - 28.6%).

### Question 5: Which F9 Booster version (v1.0, v1.1, FT, B4, B5, etc.) has the highest launch success rate?
**Answer:** **FT (Falcon 9 Full Thrust)** and **B5** have the highest launch success rates. FT has 16 successful launches out of 24 (66.7% success rate) and B5 has 100% success rate (1/1 in dataset), while B4 has 6 successes out of 11 (54.5% success rate).

```python
# Verification code for Q1 to Q5
print('--- Q1: Total Successful Launches by Site ---')
print(spacex_df[spacex_df['class'] == 1]['Launch Site'].value_counts())

print('\n--- Q2: Success Rate by Site ---')
print(spacex_df.groupby('Launch Site')['class'].agg(['count', 'sum', 'mean']))

print('\n--- Q3 & Q4: Success Rate by Payload Range ---')
spacex_df['Payload_Bin'] = pd.cut(spacex_df['Payload Mass (kg)'], bins=[0, 2500, 5000, 7500, 10000])
print(spacex_df.groupby('Payload_Bin', observed=False)['class'].agg(['count', 'sum', 'mean']))

print('\n--- Q5: Success Rate by Booster Version Category ---')
print(spacex_df.groupby('Booster Version Category')['class'].agg(['count', 'sum', 'mean']))
```
